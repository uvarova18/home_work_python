Посчитаем следующие показатели: 
    • total_spent - сумма всех трат
    • addons_spent - сумма на дополнительные услуги и аксессуары
    • orders_count - число заказов

```python
electronic_devices_sales_stats = (
    electronic_devices_sales
        .groupby("Customer ID")
        .agg (
            total_spent = ("Total Price", "sum"),
            addons_spent = ("Add-on Total", "sum"),
            orders_count = ("Customer ID","count"),
            preferred_payment=('Payment Method', get_preferred_payment)
        ))
```

Функция для поиска самого частого способа оплаты. При равной частоте способов оплаты берем первый по алфавиту:

```python
def get_preferred_payment(series):
    sorted_method = series.sort_values()
    counts = sorted_method.value_counts()
    return counts.index[0]
```

```python
electronic_devices_sales_stats = electronic_devices_sales_stats.sort_values(by='total_spent', ascending=False)
```
