Импортирую модуль JSON для возможности использования метода loads, который преобразует json в словарь.

```python
import json    
```

Создаю пустой словарь.

```python
purchases = {} 
```

Открываю файл для чтения и запускаю цикл, который построчно преобразует и складывает в переменную dict пары ключ-значение. 
Записываю значения пар в словарь.

```python
with open('purchase_log.txt', 'r') as f:
    for line in f:
        dict_ = json.loads(line) 
        purchases[dict_['user_id']] = dict_['category']
```
